# 1. Realiza un algoritmo que transforme numeros decimales a numeros binarios

```python
numero = float(input("Ingrese un número decimal para convertir a binario: ").replace(',', '.'))


parte_entera = int(numero)
parte_decimal = numero - parte_entera


binario_entero = ""
if parte_entera == 0:
    binario_entero = "0"
else:
    temp_entera = parte_entera
    while temp_entera > 0:
        residuo = temp_entera % 2
        binario_entero = str(residuo) + binario_entero
        temp_entera = temp_entera // 2


binario_decimal = ""
limite = 0

while parte_decimal > 0 and limite < 10:
    parte_decimal = parte_decimal * 2
    bit = int(parte_decimal)
    binario_decimal = binario_decimal + str(bit)
    parte_decimal = parte_decimal - bit
    limite = limite + 1


if binario_decimal != "":
    resultado = binario_entero + "." + binario_decimal
else:
    resultado = binario_entero

print("El número binario es:", resultado)
```

# 2. Realiza un algoritmo que transforme numeros binarios a numeros decimales

```python

binario = input("Ingrese el número binario (ejemplo 111101.1010100 o 1100.1): ").replace(',', '.')


parte_entera_str = ""
parte_decimal_str = ""
encontro_punto = False

for caracter in binario:
    if caracter == ".":
        encontro_punto = True
    elif not encontro_punto:
        parte_entera_str = parte_entera_str + caracter
    else:
        parte_decimal_str = parte_decimal_str + caracter

entero_decimal = 0
longitud = len(parte_entera_str)
posicion = 0

for bit in parte_entera_str:
    exponente = longitud - 1 - posicion
    entero_decimal = entero_decimal + int(bit) * (2 ** exponente)
    posicion = posicion + 1


decimal_decimal = 0.0
posicion_decimal = 1

for bit in parte_decimal_str:
    if bit == "1":
        decimal_decimal = decimal_decimal + (1 / (2 ** posicion_decimal))
    posicion_decimal = posicion_decimal + 1


if decimal_decimal == 0:
    resultado_final = entero_decimal
else:
    resultado_final = entero_decimal + decimal_decimal

print("El número en sistema decimal es:", resultado_final)
```

# 3. Realiza un algoritmo que transforme numeros romanos a numeros decimales

```python

romano = input("Ingrese un número romano (ej. I, C, XD, CXXIII): ").upper()

decimal = 0
longitud = len(romano)


for i in range(longitud):
    letra = romano[i]
    valor_actual = 0


    if letra == 'I':
        valor_actual = 1
    elif letra == 'V':
        valor_actual = 5
    elif letra == 'X':
        valor_actual = 10
    elif letra == 'L':
        valor_actual = 50
    elif letra == 'C':
        valor_actual = 100
    elif letra == 'D':
        valor_actual = 500
    elif letra == 'M':
        valor_actual = 1000


    valor_siguiente = 0
    if i + 1 < longitud:
        siguiente_letra = romano[i + 1]
        if siguiente_letra == 'I':
            valor_siguiente = 1
        elif siguiente_letra == 'V':
            valor_siguiente = 5
        elif siguiente_letra == 'X':
            valor_siguiente = 10
        elif siguiente_letra == 'L':
            valor_siguiente = 50
        elif siguiente_letra == 'C':
            valor_siguiente = 100
        elif siguiente_letra == 'D':
            valor_siguiente = 500
        elif siguiente_letra == 'M':
            valor_siguiente = 1000

    if valor_actual < valor_siguiente:
        decimal = decimal - valor_actual
    else:
        decimal = decimal + valor_actual

print("El número en sistema decimal es:", decimal)
```

# 4. Realiza un algoritmo que transforme numeros decimales a numeros romanos

```python
numero = int(input("Ingrese un número entero para convertir a romano: "))

romano = ""


while numero >= 1000:
    romano = romano + "M"
    numero = numero - 1000

while numero >= 900:
    romano = romano + "CM"
    numero = numero - 900

while numero >= 500:
    romano = romano + "D"
    numero = numero - 500

while numero >= 400:
    romano = romano + "CD"
    numero = numero - 400

while numero >= 100:
    romano = romano + "C"
    numero = numero - 100

while numero >= 90:
    romano = romano + "XC"
    numero = numero - 90

while numero >= 50:
    romano = romano + "L"
    numero = numero - 50

while numero >= 40:
    romano = romano + "XL"
    numero = numero - 40

while numero >= 10:
    romano = romano + "X"
    numero = numero - 10

while numero >= 9:
    romano = romano + "IX"
    numero = numero - 9

while numero >= 5:
    romano = romano + "V"
    numero = numero - 5

while numero >= 4:
    romano = romano + "IV"
    numero = numero - 4

while numero >= 1:
    romano = romano + "I"
    numero = numero - 1

print("El número en sistema romano es:", romano)
```

# 5. Realiza un algoritmo que valide si una frase es un palindromo

```python
#primero recordar que un palindromo es una palabra o frase cuyas letras están dispuestas de tal manera que resulta la misma leída de izquierda a derecha que de derecha a izquierda

texto = input("Ingrese una palabra o frase: ")
textolimpio = texto.lower().replace(" ", "")

longitud = len(textolimpio)

textoinvertido = ""
for i in range(longitud):
    posicion_inversa = longitud - 1 - i
    textoinvertido = textoinvertido + textolimpio[posicion_inversa]


if textolimpio == textoinvertido:
    print("La palabra/frase ES un palíndromo")
else:
    print("La palabra/frase NO es un palíndromo")
```

# 6. Realiza un algoritmo que cuente la cantidad de ocurrencias de cada caracter en una frase

```python
texto = input("Ingrese una frase: ")
contador = {}

for indice in range(len(texto)):
    caracter = texto[indice]
    if caracter in contador:
        contador[caracter] = contador[caracter] + 1
    else:
        contador[caracter] = 1

print(contador)
```

# 8. Realiza un algoritmo que valide si un email es valido
### El texto debe tener un nombre de usuario sin caracteres especiales, el símbolo @, un dominio y una extensión como .com

```python
#antes de todo toca saber que un correo se compone de "Usuario@dominio"

email = input("Ingrese su correo electrónico: ")

es_valido = True

if email.count("@") != 1:
    es_valido = False
else:

    partes = email.split("@")
    usuario = partes[0]
    dominio_completo = partes[1]


    if len(usuario) == 0:
        es_valido = False
    else:
        for caracter in usuario:
            if not caracter.isalnum() and caracter != ".":
                es_valido = False

    if "." not in dominio_completo:
        es_valido = False
    else:
        partes_dominio = dominio_completo.split(".")
        nombre_dominio = partes_dominio[0].lower()
        extension = partes_dominio[1].lower()


        dominios_permitidos = ["gmail", "outlook", "yahoo", "hotmail"]

        if nombre_dominio not in dominios_permitidos:
            es_valido = False

        if extension != "com":
            es_valido = False


if es_valido:
    print("El correo es VÁLIDO")
else:
    print("El correo NO es válido")
```

# 9. Compresor y Descompresor de Cadenas por RLE con Anidamiento Sencillo

### Implementar la codificación Run-Length Encoding (RLE) expandida
Convierte la cadena "AAABBCDDDD" en "A3B2C1D4"

```python
cadena = input("Ingrese la cadena a comprimir (ej. AAABBCDDDD): ")

if len(cadena) == 0:
    print("La cadena está vacía.")
else:
    resultado = ""
    contador = 1


    for i in range(len(cadena) - 1):
        if cadena[i] == cadena[i + 1]:
            contador = contador + 1
        else:
            resultado = resultado + cadena[i] + str(contador)
            contador = 1


    resultado = resultado + cadena[-1] + str(contador)

    print("Cadena comprimida:", resultado)
```

# 12. Implemente un algoritmo que como entrada reciba una lista de numeros, y un numero T objetivo. Como salida el algoritmo debe encontrar grupos de 3 numeros que sumen el numero T objetivo

```python
#entonces se ponen 2 input para poder agregar la lista de numeros y luego el numero T objetivo
entrada_numeros = input("Ingrese la lista de números separados por espacio o coma (ej: 0 2 18 10 8 12 -2 22): ")
objetivo = int(input("Ingrese el número objetivo T: "))


numeros = []
for num in entrada_numeros.replace(',', ' ').split():
    numeros.append(int(num))

grupos = []


for i in range(len(numeros)):
    for j in range(i + 1, len(numeros)):
        for k in range(j + 1, len(numeros)):
            if numeros[i] + numeros[j] + numeros[k] == objetivo:
                grupos.append([numeros[i], numeros[j], numeros[k]])


print("Grupos de 3 números que suman", objetivo, ":", grupos)
```

# 14. Implemente un algoritmo que transforme letras a codigo morse

```python

morse = {
    'a': '.-',
    'b': '-...',
    'c': '-.-.',
    'd': '-..',
    'e': '.',
    'f': '..-.',
    'g': '--.',
    'h': '....',
    'i': '..',
    'j': '.---',
    'k': '-.-',
    'l': '.-..',
    'm': '--',
    'n': '-.',
    'o': '---',
    'p': '.--.',
    'q': '--.-',
    'r': '.-.',
    's': '...',
    't': '-',
    'u': '..-',
    'v': '...-',
    'w': '.--',
    'x': '-..-',
    'y': '-.--',
    'z': '--..',
    ' ': '/',
}


texto = input("Ingrese una palabra o frase: ").lower()

resultado = ""


for caracter in texto:
    if caracter in morse:
        resultado = resultado + morse[caracter] + " "


print("El mensaje en código Morse es:", resultado)
```
