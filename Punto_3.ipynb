# Punto 3 - Abordaje de avion particular



```python
import random
import matplotlib.pyplot as plt
```

```python
def simular_abordaje(n_asientos=100):

    asientos = [False] * (n_asientos + 1)

    # pasajero 1
    primer_asiento = random.randint(1, n_asientos)
    asientos[primer_asiento] = True

    # Pasajero del 2 al 99
    for pasajero in range(2, n_asientos):
        if not asientos[pasajero]:
            asientos[pasajero] = True
        else:
            # Selecciona un asiento disponible al azar
            libres = [i for i in range(1, n_asientos + 1) if not asientos[i]]
            eleccion = random.choice(libres)
            asientos[eleccion] = True

    # posición del pasajero 100
    if not asientos[n_asientos]:
        return n_asientos
    else:
        libres = [i for i in range(1, n_asientos + 1) if not asientos[i]]
        return libres[0]
```

```python
# Simulación
num_simulaciones = 100000
resultados = [simular_abordaje(100) for _ in range(num_simulaciones)]

# Cálculo de la probabilidad
exitos = resultados.count(100)
probabilidad = exitos / num_simulaciones

print(f"Probabilidad estimada del pasajero 100 en su asiento: {probabilidad:.4f} ({probabilidad * 100:.2f}%)")
```

```python
# Conteo de frecuencia del pasajero 100
frecuencias = {i: resultados.count(i) for i in range(1, 101)}

#  gráfico de barras
plt.figure(figsize=(10, 5))
plt.bar(frecuencias.keys(), frecuencias.values(), color='skyblue', edgecolor='black')
plt.title("Distribución del asiento asignado al Pasajero 100")
plt.xlabel("Número de Asiento")
plt.ylabel("Frecuencia de simulaciones")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
